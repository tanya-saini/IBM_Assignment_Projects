```python
#Q1. Find the Product.
def list_prdt(l):
  prdt=1
  for i in l:
    prdt=prdt*i
  return prdt

l=[1,2,3,4,5]
list_prdt(l)
```

```python
#Q2. Find the Sum.
def list_add(l):
  add=0
  for i in l:
    add=add+i
  return add

l=[1,2,3,4,5]
list_add(l)
```

```python
#Q3. Count Occurrences
def counting(l,x):
  return l.count(x)

l=[1,1,1,2,4]
x=1
counting(l,x)
```

```python
#Q4. Even Odd
def even_odd(l):
  even_sum=sum(i for i in l if i%2==0)
  odd_sum=sum(i for i in l if i%2!=0)
  return [even_sum,odd_sum]

l=[1,2,3,4,5]
even_odd(l)
```

```python
#Q5. Find whether the number is present or not
def present(l,x):
  if x in l:
    return "yes"
  else:
    return "no"

l=[1,2,3,4,5]
x=6
present(l,x)
```

```python
#Q6. Higher Age
def highAge(l):
  for i in l:
    if i<18:
      l.remove(i)
  return l

l=[11,23,45,3,76,49]
highAge(l)
```

```python
#Q7. Increment the List Elements
def inc_list(l):
  return [i+1 for i in l]

l=[1,3,5,7,9]
inc_list(l)
```

```python
#Q8. Pass or Fail
def marks(l):
  for i in l:
    if i<32:
      return "no"
    return "yes"

l=[11,33,45,3,76,49]
marks(l)
```

```python
#Q9. Unique Color Shirt
def uniq_shirt(l):
  ans=[]
  for i in l:
    if l.count(i)==1:
      ans.append(i)
  return len(ans)

l=[2,3,4,1,3,2]
uniq_shirt(l)
```

```python
#Q10. Min and Max
def min_max(l):
  return [min(l),max(l)]

l=[1,2,3,4,5]
min_max(l)
```
