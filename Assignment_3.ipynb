```python
# Q1. Count Characters
def countChar(S):
    count_A = S.count('A')
    count_D = S.count('D')
    return [count_A, count_D]

S="AaBdDDDe"
countChar(S)
```

```python
#Q2. Count the Heads
def countHeads(s):
  lst=[]
  for i in set(s):
    if s.count(i)>1:
      lst.append(f'{i}{s.count(i)}')
  return lst

s='prepbytes'
countHeads(s)
```

```python
#Q3. Count the Vowels
def count_vowels(s):
  vow='aeiou'
  countVow=0
  for i in vow:
    countVow+=s.count(i)
  return countVow

s="sun rises in the east"
count_vowels(s)
```

```python
#Q4. Concatenate the Strings
def concat_str(s1,s2):
  return s1+s2

s1="Hello"
s2="World"
concat_str(s1,s2)
```

```python
#Q5. Find Length
def find_length(s):
  return len(s)

s="Python"
find_length(s)
```

```python
#Q6. Find the Winner
def winner(s):
  count_a=s.count('A')
  count_b=s.count('B')
  if count_a>count_b:
    return 'A'
  elif count_a<count_b:
    return 'B'
  else:
    return 'Draw'

s="AABBBBAABBA"
winner(s)
```

```python
#Q7. Join Strings
def join_str(s1,s2):
  return s1+' '+s2

s1="Hello"
s2="World"
join_str(s1,s2)
```

```python
#Q8. Plaindrome Check
def palindrome_check(s):
  if s==s[::-1]:
    return 'Yes'
  else:
    return 'No'

s="madam"
palindrome_check(s)
```

```python
#Q9. Reverse the String
def rev_str(s):
  rev=""
  for i in s:
    rev=i+rev
  return rev

s="I am utkarsh raj"
rev_str(s)
```

```python
#Q10. Match the Strings
def match_str(s1,s2):
  if s1==s2:
    return 'Yes'
  else:
    return 'No'

s1="Hello"
s2="World"
match_str(s1,s2)
```

```python
#Q11. String Replace
def replace_str(s,pattern,text):
  return s.replace(pattern,text)

s="Hi, I am You."
pattern="You"
text="Prepbytes"
replace_str(s,pattern,text)
```

```python
#Q12. Split the String
def split_str(s):
  lst=s.split(" ")
  for i in lst:
    print(i)
  return

s="Sun rises in the East"
split_str(s)
```

```python
#Q13. Count the Vowels and Consonants
def counting(s):
  vow="aeiou"
  count_vow=0
  count_con=len(s)-count_vow
  for i in vow:
    count_vow+=s.lower().count(i)
  return [count_vow,count_con]

s="helloIndia"
counting(s)
```
