```python
# Q1. Find Grades
def find_grade(x):
  if x<=10:
    print("E")
  elif x<=20:
    print("D")
  elif x<=30:
    print("C")
  elif x<=40:
    print("B")
  elif x<=50:
    print("A")

num=int(input("Enter the marks: "))
find_grade(num)
```

```python
# Q2. Get Value
def get_value(n):
  if n.upper() in ['P','E','D','Z']:
    if n.upper()=='E':
      print('Expert Coder')
    elif n.upper()=='Z':
      print("Zenith")
    elif n.upper()=='D':
      print("Data Structure")
    else:
      print("PrepBytes")

c=input("enter a alphabet  ")
get_value(c)
```

```python
# Q3. Find the maximum out of three numbers.
def find_max(a,b,c):
  if a>b and a>c:
    print(a,' is max')
  elif b>c and b>a:
    print(b,' is max')
  else:
    print(c," is max")

find_max(2,5,4)
```

```python
# Q4. Find Second Smallest.
def find_2nd_min(a,b,c):
  if a<b and a<c:
    if b<c:
      print(b,' is 2nd min')
    else:
      print(c,' is 2nd min')
  else:
    print(a,' is 2nd min')

find_2nd_min(2,5,4)
```

```python
# Q5. Check whether the triangle is Acute or Obtuse
def triangle(a,b,c):
  if a>90 or b>90 or c>90:
    print("Obtuse Triangle")
  else:
    print("Acute Triangle")

triangle(60,100,20)
```
