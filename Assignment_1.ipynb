```python
# Q1. Add two Numbers
def add2num(a,b):
  return a+b

add2num(2,3)
```

```python
# Q2. Find if the conditions are obeyed or not?
def is_valid(a,b):
  return a<10 and a>b

is_valid(4,1)
```

```python
# Q3. Check the conditions
def check(a,b):
  return a%10==0 or b%10==0

check(10,10)
```

```python
# Q4. Find the first digit of a 4 digit number
def first_digit(a):
  return a//1000

first_digit(6534)
```

```python
# Q5.Find the last digit of a 4 digit number
def last_digit(a):
  return a%10

last_digit(6534)
```

```python
# Q6. Find the remainder
def remainder(a,b):
  return b%a

remainder(2,9)
```

```python
# Q7. Multiply two Numbers
def multiple(a,b):
  return b*a

multiple(2,9)
```

```python
# Q8. Marks Calculator
def summ(a,b,c):
  return a+b+c

def avg(a,b,c):
  return summ(a,b,c)/3

print(summ(20,50,100), f'{avg(20,50,100):.2f}')
```
